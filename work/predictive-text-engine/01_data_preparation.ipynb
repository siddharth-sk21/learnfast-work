```python
from nltk.corpus import gutenberg
from pathlib import Path

selected_books = [
    'austen-emma.txt',
    'austen-persuasion.txt',
    'austen-sense.txt',
    'melville-moby_dick.txt',
    'carroll-alice.txt',
    'chesterton-ball.txt',
    'chesterton-thursday.txt'
]

all_text = []

for book in selected_books:
    text = gutenberg.raw(book)
    all_text.append(text)

corpus = "\n\n".join(all_text)

data_dir = Path("data")
data_dir.mkdir(exist_ok=True)

with open(data_dir / "corpus.txt", "w", encoding="utf-8") as f:
    f.write(corpus)

print("Corpus created successfully!")
```

```python
import re
from collections import Counter

words = re.findall(r"[a-z']+", corpus.lower())

sentences = re.split(r"[.!?]+", corpus)

print("Dataset Statistics")
print("------------------------")
print("Total characters:", len(corpus))
print("Total words:", len(words))
print("Total sentences:", len(sentences))
print("Unique words:", len(set(words)))

print("\nTop 15 frequent words:")
print(Counter(words).most_common(15))
```
